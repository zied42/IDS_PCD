```python
# Cell C — Feature Importance selection for ALL algorithms
# Note: DT, RF, XGBoost, LightGBM have native importances; KNN uses SelectKBest fallback

fi_results  = {}
fi_clfs     = {}

for name in ALGORITHMS:
    print(f"\n{'='*50}\nFeat. Importance — {name}")
    clf_base = optuna_best_clf[name]
    bp       = optuna_studies[name].best_params

    # Get feature importances
    if name in ("Decision Tree", "Random Forest", "XGBoost", "LightGBM"):
        importances = clf_base.feature_importances_
        sorted_idx  = np.argsort(importances)[::-1]
        top_feats   = [X_train.columns[i] for i in sorted_idx[:PSO_TOP_N]]
    else:  # KNN — no native importances, fall back to f_classif
        sel = SelectKBest(f_classif, k=PSO_TOP_N).fit(X_train, y_train)
        top_feats = X_train.columns[sel.get_support()].tolist()
    print(f"  Top {PSO_TOP_N} features selected")

    X_tr_fi = X_train[top_feats]
    X_va_fi = X_val[top_feats]
    X_te_fi = X_test[top_feats]

    start = time.time()
    if name == "Decision Tree":
        clf = DecisionTreeClassifier(**bp, random_state=RANDOM_STATE)
    elif name == "Random Forest":
        clf = RandomForestClassifier(**bp, n_jobs=-1, random_state=RANDOM_STATE)
    elif name == "KNN":
        clf = KNeighborsClassifier(**bp, n_jobs=-1)
    elif name == "XGBoost":
        clf = XGBClassifier(**bp, tree_method="hist", device="cpu", random_state=RANDOM_STATE)
        clf.fit(X_tr_fi, y_train, eval_set=[(X_va_fi, y_val)],
                early_stopping_rounds=50, verbose=False)
        train_time = time.time() - start
        fi_clfs[name] = clf
        proba = clf.predict_proba(X_te_fi)[:, 1]
        m = ids_evaluate(f"{name}+FI", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
        m["Train Time (s)"] = round(train_time, 2)
        fi_results[name] = m
        print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")
        continue
    elif name == "LightGBM":
        clf = lgb.LGBMClassifier(**bp, random_state=RANDOM_STATE, verbose=-1)
        clf.fit(X_tr_fi, y_train, eval_set=[(X_va_fi, y_val)],
                callbacks=[lgb.early_stopping(50, verbose=False)])
        train_time = time.time() - start
        fi_clfs[name] = clf
        proba = clf.predict_proba(X_te_fi)[:, 1]
        m = ids_evaluate(f"{name}+FI", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
        m["Train Time (s)"] = round(train_time, 2)
        fi_results[name] = m
        print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")
        continue

    clf.fit(X_tr_fi, y_train)
    train_time = time.time() - start
    fi_clfs[name] = clf
    proba = clf.predict_proba(X_te_fi)[:, 1]
    m = ids_evaluate(f"{name}+FI", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
    m["Train Time (s)"] = round(train_time, 2)
    fi_results[name] = m
    print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")

print("\nFeature Importance selection done for all algorithms.")
```

```python
# Cell F — PSO feature selection for ALL algorithms
import pyswarms as ps
from sklearn.model_selection import train_test_split as tts

print(f"Subsampling {SAMPLE_PSO} rows for PSO evaluation speed...")
_, X_tr_pso_df, _, y_tr_pso = tts(
    X_train, y_train.values,
    test_size=SAMPLE_PSO / len(X_train),
    stratify=y_train.values,
    random_state=RANDOM_STATE
)
X_tr_pso = X_tr_pso_df.values
X_va_pso = X_val.values
y_va_pso = y_val.values
n_feat   = X_tr_pso.shape[1]

def make_pso_objective(algo_name, bp):
    def pso_objective(particles):
        costs = []
        for particle in particles:
            mask = particle > 0.5
            if mask.sum() == 0:
                costs.append(1.0); continue
            if algo_name == "Decision Tree":
                clf = DecisionTreeClassifier(**bp, random_state=RANDOM_STATE)
            elif algo_name == "Random Forest":
                clf = RandomForestClassifier(**bp, n_jobs=-1, random_state=RANDOM_STATE)
            elif algo_name == "KNN":
                clf = KNeighborsClassifier(**bp, n_jobs=-1)
            elif algo_name == "XGBoost":
                clf = XGBClassifier(**bp, tree_method="hist", device="cpu",
                                    n_estimators=100, random_state=RANDOM_STATE, verbosity=0)
            elif algo_name == "LightGBM":
                clf = lgb.LGBMClassifier(**bp, n_estimators=100,
                                         random_state=RANDOM_STATE, verbose=-1)
            clf.fit(X_tr_pso[:, mask], y_tr_pso)
            proba = clf.predict_proba(X_va_pso[:, mask])[:, 1]
            costs.append(1 - balanced_accuracy_score(y_va_pso, (proba>=0.5).astype(int)))
        return np.array(costs)
    return pso_objective

pso_results  = {}
pso_clfs     = {}
pso_features_all = {}

for name in ALGORITHMS:
    print(f"\n{'='*50}\nPSO — {name}")
    bp = optuna_studies[name].best_params
    options   = {"c1": 0.5, "c2": 0.3, "w": 0.9}
    optimizer = ps.single.GlobalBestPSO(
        n_particles=PSO_PARTICLES, dimensions=n_feat,
        options=options, bounds=(np.zeros(n_feat), np.ones(n_feat))
    )
    best_cost, best_pos = optimizer.optimize(make_pso_objective(name, bp), iters=PSO_ITERS)
    pso_mask  = best_pos > 0.5
    top_feats = [f for f, s in zip(X_train.columns, pso_mask) if s]
    pso_features_all[name] = top_feats
    print(f"  PSO selected {pso_mask.sum()} / {n_feat} features")

    X_tr_p = X_train[top_feats]; X_va_p = X_val[top_feats]; X_te_p = X_test[top_feats]
    start = time.time()
    if name == "Decision Tree":
        clf = DecisionTreeClassifier(**bp, random_state=RANDOM_STATE)
        clf.fit(X_tr_p, y_train)
    elif name == "Random Forest":
        clf = RandomForestClassifier(**bp, n_jobs=-1, random_state=RANDOM_STATE)
        clf.fit(X_tr_p, y_train)
    elif name == "KNN":
        clf = KNeighborsClassifier(**bp, n_jobs=-1)
        clf.fit(X_tr_p, y_train)
    elif name == "XGBoost":
        clf = XGBClassifier(**bp, tree_method="hist", device="cpu", random_state=RANDOM_STATE)
        clf.fit(X_tr_p, y_train, eval_set=[(X_va_p, y_val)],
                early_stopping_rounds=50, verbose=False)
    elif name == "LightGBM":
        clf = lgb.LGBMClassifier(**bp, random_state=RANDOM_STATE, verbose=-1)
        clf.fit(X_tr_p, y_train, eval_set=[(X_va_p, y_val)],
                callbacks=[lgb.early_stopping(50, verbose=False)])

    train_time = time.time() - start
    pso_clfs[name] = clf
    proba = clf.predict_proba(X_te_p)[:, 1]
    m = ids_evaluate(f"{name}+PSO", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
    m["Train Time (s)"] = round(train_time, 2)
    pso_results[name] = m
    print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")

with open("models/ml/pso_features_all.pkl", "wb") as f:
    pickle.dump(pso_features_all, f)
print("\nPSO done for all algorithms.")
```

```python
# Cell G — SMOTE for ALL algorithms
from sklearn.model_selection import train_test_split as tts

print(f"Subsampling {SAMPLE_SMOTE} rows then applying SMOTE...")
_, X_sm_df, _, y_sm_raw = tts(
    X_train, y_train.values,
    test_size=SAMPLE_SMOTE / len(X_train),
    stratify=y_train.values, random_state=RANDOM_STATE
)
smote      = SMOTE(random_state=RANDOM_STATE)
X_sm, y_sm = smote.fit_resample(X_sm_df.values, y_sm_raw)
print(f"Before SMOTE: {np.bincount(y_sm_raw)}  →  After: {np.bincount(y_sm)}")

smote_results = {}
smote_clfs    = {}

for name in ALGORITHMS:
    print(f"\n{'='*50}\nSMOTE — {name}")
    bp = optuna_studies[name].best_params
    start = time.time()

    if name == "Decision Tree":
        clf = DecisionTreeClassifier(**bp, random_state=RANDOM_STATE)
    elif name == "Random Forest":
        clf = RandomForestClassifier(**bp, n_jobs=-1, random_state=RANDOM_STATE)
    elif name == "KNN":
        clf = KNeighborsClassifier(**bp, n_jobs=-1)
    elif name == "XGBoost":
        clf = XGBClassifier(**bp, tree_method="hist", device="cpu", random_state=RANDOM_STATE)
    elif name == "LightGBM":
        clf = lgb.LGBMClassifier(**bp, random_state=RANDOM_STATE, verbose=-1)

    clf.fit(X_sm, y_sm)   # no early stopping — SMOTE set has no separate val
    train_time = time.time() - start
    smote_clfs[name] = clf
    proba = clf.predict_proba(X_test)[:, 1]
    m = ids_evaluate(f"{name}+SMOTE", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
    m["Train Time (s)"] = round(train_time, 2)
    smote_results[name] = m
    print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")

print("\nSMOTE done for all algorithms.")
```

```python
# Cell H — Class imbalance correction for ALL algorithms
# XGBoost/LightGBM → scale_pos_weight
# DT/RF           → class_weight="balanced"
# KNN             → no native support, skip (mark N/A)

n_benign = int((y_train == 0).sum())
n_attack = int((y_train == 1).sum())
spw      = n_benign / n_attack
print(f"Benign: {n_benign:,}  |  Attack: {n_attack:,}  |  scale_pos_weight: {spw:.4f}")

spw_results = {}
spw_clfs    = {}

for name in ALGORITHMS:
    print(f"\n{'='*50}\nClass Balance — {name}")
    bp = optuna_studies[name].best_params
    start = time.time()

    if name == "Decision Tree":
        clf = DecisionTreeClassifier(**bp, class_weight="balanced", random_state=RANDOM_STATE)
        clf.fit(X_train, y_train)
    elif name == "Random Forest":
        clf = RandomForestClassifier(**bp, class_weight="balanced",
                                     n_jobs=-1, random_state=RANDOM_STATE)
        clf.fit(X_train, y_train)
    elif name == "KNN":
        print("  KNN has no native class-weight support — using plain Optuna model as proxy.")
        clf = optuna_best_clf["KNN"]
        train_time = 0.0
        proba = clf.predict_proba(X_test)[:, 1]
        m = ids_evaluate("KNN+ClassBal(N/A)", y_test.values, proba,
                         (proba>=0.5).astype(int), 0.5)
        m["Train Time (s)"] = 0.0
        spw_results[name] = m
        print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}")
        continue
    elif name == "XGBoost":
        clf = XGBClassifier(**bp, scale_pos_weight=spw,
                            tree_method="hist", device="cpu", random_state=RANDOM_STATE)
        clf.fit(X_train, y_train, eval_set=[(X_val, y_val)],
                early_stopping_rounds=50, verbose=False)
    elif name == "LightGBM":
        clf = lgb.LGBMClassifier(**bp, scale_pos_weight=spw,
                                  random_state=RANDOM_STATE, verbose=-1)
        clf.fit(X_train, y_train, eval_set=[(X_val, y_val)],
                callbacks=[lgb.early_stopping(50, verbose=False)])

    train_time = time.time() - start
    spw_clfs[name] = clf
    proba = clf.predict_proba(X_test)[:, 1]
    m = ids_evaluate(f"{name}+ClassBal", y_test.values, proba, (proba>=0.5).astype(int), 0.5)
    m["Train Time (s)"] = round(train_time, 2)
    spw_results[name] = m
    print(f"  Test Balanced Acc: {m['Balanced Acc']:.4f}  ({train_time:.1f}s)")

print("\nClass balance correction done for all algorithms.")
```

```python
# Cell I — Master summary: every algorithm × every optimization method
import os
os.makedirs("outputs/reports", exist_ok=True)

METHOD_RESULTS = {
    "Baseline":   all_results,         # from notebook 02
    "Optuna":     optuna_results,
    "Feat. Imp.": fi_results,
    "UniKB":      uni_results,
    "PCA":        pca_results,
    "PSO":        pso_results,
    "SMOTE":      smote_results,
    "ClassBal":   spw_results,
}

rows = []
for method, res_dict in METHOD_RESULTS.items():
    for algo, m in res_dict.items():
        rows.append({
            "Algorithm": algo,
            "Method":    method,
            **{k: v for k, v in m.items() if k != "Model"}
        })

master_df = pd.DataFrame(rows)
master_df = master_df.sort_values(["Algorithm", "Balanced Acc"], ascending=[True, False])
master_df.to_csv("outputs/reports/03_all_algorithms_optimization.csv", index=False)

print("MASTER RESULTS — All Algorithms × All Optimization Methods")
print("=" * 110)
cols = ["Algorithm", "Method", "Balanced Acc", "F1 Macro", "MCC",
        "Attack Recall", "Benign Recall", "FAR", "FNR", "Train Time (s)"]
print(master_df[cols].to_string(index=False))

# Best method per algorithm
print("\n\nBEST METHOD PER ALGORITHM:")
print("=" * 60)
best_per_algo = master_df.loc[master_df.groupby("Algorithm")["Balanced Acc"].idxmax()]
print(best_per_algo[["Algorithm", "Method", "Balanced Acc", "Attack Recall", "FAR"]].to_string(index=False))
```
